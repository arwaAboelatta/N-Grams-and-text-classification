**Confusion Matrix:**

```python
import numpy as np

def confusion_matrix(y_true, y_pred, num_classes=5): #initialzing an empty matrix

    matrix = np.zeros((num_classes, num_classes), dtype=int)

    for true, pred in zip(y_true, y_pred):
        matrix[true][pred] += 1

    return matrix
```

```python
def precision(matrix,num_classes=5):
    precision = np.zeros(num_classes)

    for i in range(num_classes):
        tp = matrix[i, i]
        fp = np.sum(matrix[:, i]) - tp
        precision[i] = tp / (tp + fp) if (tp + fp) > 0 else 0

    macro_precision = np.mean(precision)
    return precision, macro_precision
```

```python
def recall(matrix,num_classes=5):
    recall = np.zeros(num_classes)

    for i in range(num_classes):
        tp = matrix[i, i]
        fn = np.sum(matrix[i, :]) - tp
        recall[i] = tp / (tp + fn) if (tp + fn) > 0 else 0

    macro_recall = np.mean(recall)
    return recall, macro_recall
```

```python
def f1_score(matrix,num_classes=5):
    precision_per_class, _ = precision(matrix,num_classes)
    recall_per_class, _ = recall(matrix,num_classes)

    f1 = np.zeros(len(precision_per_class))

    for i in range(len(precision_per_class)):
        p, r = precision_per_class[i], recall_per_class[i]
        f1[i] = (2 * p * r) / (p + r) if (p + r) > 0 else 0

    macro_f1 = np.mean(f1)
    return f1, macro_f1
```

**Naive Bayes:**

```python
import numpy as np
from collections import defaultdict

class NaiveBayesClassifier:
    def __init__(self):
        self.logprior = {}
        self.loglikelihood = defaultdict(lambda: defaultdict(float))
        self.vocab = set()
        self.classes = [0, 1, 2, 3, 4]

    def train(self, documents, labels):
        N_doc = len(documents) #total no of documents in the dataset
        class_docs = {c: [] for c in self.classes}      #doc of each class
        word_counts = {c: defaultdict(int) for c in self.classes}      #words of each class
        class_counts = defaultdict(int)     #no of documents in a class

        #combining all documents to make one big document
        for doc, label in zip(documents, labels):
            class_docs[label].append(doc)
            class_counts[label] += 1
            for word in doc.split():
                word_counts[label][word] += 1
                self.vocab.add(word)

        #loop over all classes -> 0 , 1 , 2 , 3 , 4
        for c in self.classes:
            self.logprior[c] = np.log(class_counts[c] / N_doc) #calculating priors
            total_words_in_class = sum(word_counts[c].values())   #number of words in this class
            #calculating the likelihoods
            for word in self.vocab:
                count_wc = word_counts[c][word] + 1  #smoothing -> used the m-estimator with m=1
                denominator = total_words_in_class + len(self.vocab)
                self.loglikelihood[word][c] = np.log(count_wc / denominator)

    def test(self, document):
        sum_log_probs = {c: self.logprior[c] for c in self.classes}
        for word in document.split():
            if word in self.vocab:
                for c in self.classes:
                    sum_log_probs[c] += self.loglikelihood[word][c]
        return max(sum_log_probs, key=sum_log_probs.get)  #we return the class with maximum probbailty
```

**Logistic Regression**

```python
import numpy as np
import pandas as pd
from datasets import load_dataset
from collections import Counter
from sklearn.preprocessing import OneHotEncoder
from scipy.sparse import csr_matrix, hstack, vstack
from sklearn.linear_model import LogisticRegression, SGDClassifier
from sklearn.metrics import accuracy_score, classification_report
```

```python
def generate_bigrams(tokens):
    words = tokens.split("|")
    return list(zip(words, words[1:]))
```

```python
def build_bigram_vocabulary(df):
    all_bigrams = []
    for tokenized_sentence in df["tokens"]:
        all_bigrams.extend(generate_bigrams(tokenized_sentence))

    # Count bigram frequencies
    bigram_counts = Counter(all_bigrams)

    # Create vocabulary mapping each bigram to an index
    bigram_vocab = {bigram: idx for idx, bigram in enumerate(bigram_counts.keys())}

    print(f"Total unique bigrams: {len(bigram_vocab)}")
    return bigram_vocab

print("\nBuilding bigram vocabulary from training data")
bigram_vocab = build_bigram_vocabulary(df_train)
```

```python
def create_sparse_feature_matrix(df, bigram_vocab):
    rows = []
    cols = []
    data = []

    for row_idx, tokenized_sentence in enumerate(df["tokens"]):
        bigrams = generate_bigrams(tokenized_sentence)
        for bigram in bigrams:
            if bigram in bigram_vocab:
                rows.append(row_idx)
                cols.append(bigram_vocab[bigram])
                # 1 if bigram exists
                data.append(1)

    # Create sparse matrix in compressed format to save memory
    X_sparse = csr_matrix((data, (rows, cols)), shape=(len(df), len(bigram_vocab)))
    return X_sparse
```

```python
X_train_sparse = create_sparse_feature_matrix(df_train, bigram_vocab)
X_test_sparse = create_sparse_feature_matrix(df_test, bigram_vocab)
X_val_sparse = create_sparse_feature_matrix(df_validation, bigram_vocab)

print(f"Training feature matrix shape: {X_train_sparse.shape}")
print(f"Test feature matrix shape: {X_test_sparse.shape}")
print(f"Validation feature matrix shape: {X_val_sparse.shape}")
```

```python
encoder = OneHotEncoder(sparse_output=True)
y_train_sparse = encoder.fit_transform(df_train["class"].values.reshape(-1, 1))
y_test_sparse = encoder.transform(df_test["class"].values.reshape(-1, 1))
y_val_sparse = encoder.transform(df_validation["class"].values.reshape(-1, 1))
```

```python
def initialize_weights(n_features, n_classes):
    # Initialize with small random values
    W = np.random.randn(n_features, n_classes) * 0.01
    b = np.zeros(n_classes)
    return W, b
```

```python
def softmax(z):
    # Use the max trick for numerical stability
    exp_z = np.exp(z - np.max(z, axis=1, keepdims=True))
    return exp_z / np.sum(exp_z, axis=1, keepdims=True)
```

```python
def compute_loss(y_true, y_pred, W, lambda_reg):
    m = y_true.shape[0]
    # Prevent log(0)
    epsilon = 1e-15

    # cross-entropy
    data_loss = -np.mean(np.sum(y_true * np.log(y_pred + epsilon), axis=1))

    # Regularization loss
    reg_loss = (lambda_reg / (2 * m)) * np.sum(W * W)

    return data_loss + reg_loss
```

```python
def compute_gradients(X, y_true, y_pred, W, lambda_reg):
    m = y_true.shape[0]

    # Gradient of weights
    dW = (X.T.dot(y_pred - y_true) / m) + (lambda_reg * W / m)

    # Gradient of bias
    db = np.sum(y_pred - y_true, axis=0) / m

    return dW, db
```

```python
def compute_accuracy(y_true, y_pred):
    pred_classes = np.argmax(y_pred, axis=1)
    true_classes = np.argmax(y_true, axis=1)
    return np.mean(pred_classes == true_classes)
```

```python
def get_mini_batches(X, y, batch_size):
    m = X.shape[0]
    indices = np.random.permutation(m)

    for i in range(0, m, batch_size):
        end = min(i + batch_size, m)
        batch_indices = indices[i:end]
        yield X[batch_indices], y[batch_indices]
```

```python
def adjust_learning_rate(epoch, initial_lr, decay_rate=0.95, decay_step=10):
    return initial_lr * (decay_rate ** (epoch // decay_step))
```

```python
def train_logistic_regression(X_train, y_train, X_val, y_val,
                              initial_lr=0.1, epochs=50, batch_size=64,
                              lambda_reg=0.01, early_stopping_patience=5):

    n_features = X_train.shape[1]
    n_classes = y_train.shape[1]

    # Initialize weights and bias
    W, b = initialize_weights(n_features, n_classes)

    # Track metrics
    train_losses = []
    val_losses = []
    val_accuracies = []

    # For early stopping
    best_val_loss = float('inf')
    patience_counter = 0
    best_W, best_b = None, None

    print("\nStarting training...")
    print(f"Parameters: learning_rate={initial_lr}, batch_size={batch_size}, lambda_reg={lambda_reg}")

    for epoch in range(epochs):
        # Update learning rate with decay
        lr = adjust_learning_rate(epoch, initial_lr)

        # Train on mini-batches
        for X_batch, y_batch in get_mini_batches(X_train, y_train, batch_size):
            # Convert sparse to dense for computation
            X_batch_dense = X_batch.toarray()
            y_batch_dense = y_batch.toarray()

            # Forward pass
            z = X_batch_dense.dot(W) + b
            y_pred = softmax(z)

            # Compute gradients
            dW, db = compute_gradients(X_batch_dense, y_batch_dense, y_pred, W, lambda_reg)

            # Update parameters
            W -= lr * dW
            b -= lr * db

        # Compute metrics at the end of each epoch
        # Training loss
        X_train_sample = X_train[:1000].toarray()
        y_train_sample = y_train[:1000].toarray()
        train_pred = softmax(X_train_sample.dot(W) + b)
        train_loss = compute_loss(y_train_sample, train_pred, W, lambda_reg)
        train_losses.append(train_loss)

        # Validation metrics
        X_val_dense = X_val.toarray()
        y_val_dense = y_val.toarray()
        val_pred = softmax(X_val_dense.dot(W) + b)
        val_loss = compute_loss(y_val_dense, val_pred, W, lambda_reg)
        val_losses.append(val_loss)

        val_accuracy = compute_accuracy(y_val_dense, val_pred)
        val_accuracies.append(val_accuracy)

        # Print progress
        if (epoch + 1) % 5 == 0 or epoch == 0:
            print(f"Epoch {epoch+1}/{epochs}, LR: {lr:.6f}, Train Loss: {train_loss:.4f}, "
                  f"Val Loss: {val_loss:.4f}, Val Accuracy: {val_accuracy:.4f}")

        # Early stopping check
        if val_loss < best_val_loss:
            best_val_loss = val_loss
            best_W, best_b = W.copy(), b.copy()
            patience_counter = 0
        else:
            patience_counter += 1
            if patience_counter >= early_stopping_patience:
                print(f"Early stopping at epoch {epoch+1}")
                break

    # Return the best model if early stopping occurred
    if best_W is not None and best_b is not None:
        W, b = best_W, best_b

    return W, b, train_losses, val_losses, val_accuracies
```

```python
W, b, train_losses, val_losses, val_accuracies = train_logistic_regression(
    X_train_sparse, y_train_sparse, X_val_sparse, y_val_sparse,
    initial_lr=0.1, epochs=50, batch_size=64, lambda_reg=0.01
)
```

```python
def evaluate_model(X, y, W, b, dataset_name="Test"):
    X_dense = X.toarray()
    y_dense = y.toarray()

    # Predictions
    predictions = softmax(X_dense.dot(W) + b)

    # Calculate metrics
    # No regularization for evaluation
    loss = compute_loss(y_dense, predictions, W, 0)
    accuracy = compute_accuracy(y_dense, predictions)

    # Get class predictions
    pred_classes = np.argmax(predictions, axis=1)
    true_classes = np.argmax(y_dense, axis=1)

    print(f"\n{dataset_name} Set Evaluation:")
    print(f"Loss: {loss:.4f}")
    print(f"Accuracy: {accuracy:.4f}")


    return loss, accuracy, pred_classes, true_classes
```

```python
val_metrics = evaluate_model(X_val_sparse, y_val_sparse, W, b, "Validation")
test_metrics = evaluate_model(X_test_sparse, y_test_sparse, W, b, "Test")
```
